```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023,2024,2025,2026])
y1 = np.array([15,25,30,20])
y2 = np.array([17,23,56,5])
y3 = np.array([34,12,19,48])
```

```python
line_style = dict(marker='.',markersize = 20,markerfacecolor = '#1cddeb',markeredgecolor = '#1cddeb',linestyle='solid',linewidth = 4)
font_style = dict(fontsize = 20,fontweight = 'bold',color = '#1ceb4c')
```

```python
plt.title('Class Size', **font_style)
plt.xlabel("Year",**font_style)
plt.ylabel("Students",**font_style)
plt.xticks(x) #it's use for xlabel element that give by use only use remove like this, 2023.5,2024.0 
plt.tick_params(axis = "both",colors = "#ff5d05")
plt.plot(x,y1,**line_style,color = '#1c2deb')
plt.plot(x,y2,**line_style,color = '#eb1ccf')
plt.plot(x,y3,**line_style,color = '#eb1ccf')
plt.subplots_adjust(left = 0.14,bottom=0.175)
plt.grid(axis='both',linewidth=2, color='lightgray',linestyle='dotted')
plt.show()
```
